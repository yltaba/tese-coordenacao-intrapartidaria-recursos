```python
from pathlib import Path
import pandas as pd
import numpy as np
import pandas as pd
from pathlib import Path

raw_cand_path = Path().resolve().parents[1] / "data" / "raw" / "candidatos"
OUTPUT_PATH = (
    Path().resolve().parent.parent / "data" / "processed"
)


def gerar_base_candidatos(raw_cand_path):
    cols_ate_2016 = [
        "ano_eleicao",
        "sg_uf",
        "ds_cargo",
        "sg_partido",
        "nr_candidato",
        "nm_candidato",
        "nr_cpf_candidato",
        "nm_ue",
        "nm_tipo_eleicao",
        "st_candidato_inserido_urna",
        "ds_genero",
        "ds_cor_raca",
    ]

    cand98 = pd.read_csv(
        raw_cand_path / "consulta_cand_1998_BRASIL.csv",
        sep=";",
        encoding="latin1",
        dtype={"NR_CPF_CANDIDATO": str},
    )
    cand98.columns = cand98.columns.str.lower()

    cand00 = pd.read_csv(
        raw_cand_path / "consulta_cand_2000_BRASIL.csv",
        sep=";",
        encoding="latin1",
        dtype={"NR_CPF_CANDIDATO": str},
    )
    cand00.columns = cand00.columns.str.lower()

    cand02 = pd.read_csv(
        raw_cand_path / "consulta_cand_2002_BRASIL.csv",
        sep=";",
        encoding="latin1",
        dtype={"NR_CPF_CANDIDATO": str},
    )
    cand02.columns = cand02.columns.str.lower()

    cand04 = pd.read_csv(
        raw_cand_path / "consulta_cand_2004_BRASIL.csv",
        sep=";",
        encoding="latin1",
        dtype={"NR_CPF_CANDIDATO": str},
    )
    cand04.columns = cand04.columns.str.lower()

    cand06 = pd.read_csv(
        raw_cand_path / "consulta_cand_2006_BRASIL.csv",
        sep=";",
        encoding="latin1",
        dtype={"NR_CPF_CANDIDATO": str},
    )
    cand06.columns = cand06.columns.str.lower()

    cand08 = pd.read_csv(
        raw_cand_path / "consulta_cand_2008_BRASIL.csv",
        sep=";",
        encoding="latin1",
        dtype={"NR_CPF_CANDIDATO": str},
    )
    cand08.columns = cand08.columns.str.lower()

    cand10 = pd.read_csv(
        raw_cand_path / "consulta_cand_2010_BRASIL.csv",
        sep=";",
        encoding="latin1",
        dtype={"NR_CPF_CANDIDATO": str},
    )
    cand10.columns = cand10.columns.str.lower()

    cand12 = pd.read_csv(
        raw_cand_path / "consulta_cand_2012_BRASIL.csv",
        sep=";",
        encoding="latin1",
        dtype={"NR_CPF_CANDIDATO": str},
    )
    cand12.columns = cand12.columns.str.lower()

    cand14 = pd.read_csv(
        raw_cand_path / "consulta_cand_2014_BRASIL.csv",
        sep=";",
        encoding="latin1",
        dtype={"NR_CPF_CANDIDATO": str},
    )
    cand14.columns = cand14.columns.str.lower()
    cand14 = cand14.drop_duplicates(
        subset=[
            "ano_eleicao",
            "sg_uf",
            "ds_cargo",
            "nr_candidato",
            "nm_ue",
            "nm_tipo_eleicao",
            "st_candidato_inserido_urna",
            "ds_genero",
            "ds_cor_raca",
        ],
        keep="first",
    )

    cand16 = pd.read_csv(
        raw_cand_path / "consulta_cand_2016_BRASIL.csv",
        sep=";",
        encoding="latin1",
        dtype={"NR_CPF_CANDIDATO": str},
    )
    cand16.columns = cand16.columns.str.lower()
    cand16 = cand16.loc[
        ~cand16["ds_situacao_candidato_tot"].isin(["Falecimento", "Renúncia"])
    ].copy()

    cols_desde_2018 = [
        "ano_eleicao",
        "sg_uf",
        "ds_cargo",
        "sg_partido",
        "nr_candidato",
        "nm_candidato",
        "nr_cpf_candidato",
        "nm_ue",
        "nm_tipo_eleicao",
        "ds_situacao_candidatura",
        "ds_genero",
        "ds_cor_raca",
    ]

    cand18 = pd.read_csv(
        raw_cand_path / "consulta_cand_2018_BRASIL.csv",
        sep=";",
        encoding="latin1",
        dtype={"NR_CPF_CANDIDATO": str},
    )
    cand18.columns = cand18.columns.str.lower()
    cand20 = pd.read_csv(
        raw_cand_path / "consulta_cand_2020_BRASIL.csv",
        sep=";",
        encoding="latin1",
        dtype={"NR_CPF_CANDIDATO": str},
    )
    cand20.columns = cand20.columns.str.lower()
    cand22 = pd.read_csv(
        raw_cand_path / "consulta_cand_2022_BRASIL.csv",
        sep=";",
        encoding="latin1",
        dtype={"NR_CPF_CANDIDATO": str},
    )
    cand22.columns = cand22.columns.str.lower()

    candidatos_ate_2016 = pd.concat(
        [
            cand98[cols_ate_2016],
            cand00[cols_ate_2016],
            cand02[cols_ate_2016],
            cand04[cols_ate_2016],
            cand06[cols_ate_2016],
            cand08[cols_ate_2016],
            cand10[cols_ate_2016],
            cand12[cols_ate_2016],
            cand14[cols_ate_2016],
            cand16[cols_ate_2016],
        ],
        axis=0,
    )
    candidatos_desde_2018 = pd.concat(
        [cand18[cols_desde_2018], cand20[cols_desde_2018], cand22[cols_desde_2018]],
        axis=0,
    )
    candidatos = pd.concat([candidatos_ate_2016, candidatos_desde_2018], axis=0)

    return candidatos


def processar_salvar_candidatos(raw_cand_path):

    print("Gerando bases de candidatos...")
    candidatos = gerar_base_candidatos(raw_cand_path)

    candidatos["ds_cargo"] = candidatos["ds_cargo"].str.upper()

    candidatos["nm_municipio"] = np.where(
        candidatos["ds_cargo"].isin(["VEREADOR", "PREFEITO"]),
        candidatos["nm_ue"],
        candidatos["sg_uf"],
    )
    candidatos = candidatos.drop("nm_ue", axis=1)
    
    candidatos['nr_candidato'] = candidatos['nr_candidato'].astype(str)

    # export
    candidatos_filtrado = candidatos.loc[
        (~candidatos["nr_cpf_candidato"].isin(["-1", "-4", "00000000000"]))
        & (candidatos["nm_tipo_eleicao"] == "ELEIÇÃO ORDINÁRIA")
    ].copy()

    candidatos_filtrado = candidatos_filtrado.drop("nm_tipo_eleicao", axis=1)

    return candidatos


# Execução
candidatos = processar_salvar_candidatos(raw_cand_path)
```

```python
candidatos.to_parquet(OUTPUT_PATH / "candidatos.parquet", index=False)
```
