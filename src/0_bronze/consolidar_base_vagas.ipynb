```python
import pandas as pd
from pathlib import Path


RAW_DATA_PATH = Path().resolve().parents[1] / "data" / "raw" / "vagas"
PROCESSED_DATA_PATH = Path().resolve().parents[1] / "data" / "processed"

vagas98 = pd.read_csv(
    RAW_DATA_PATH / "consulta_vagas_1998_BRASIL.csv", sep=";", encoding="latin1"
)

vagas02 = pd.read_csv(
    RAW_DATA_PATH / "consulta_vagas_2002_BRASIL.csv", sep=";", encoding="latin1"
)
vagas06 = pd.read_csv(
    RAW_DATA_PATH / "consulta_vagas_2006_BRASIL.csv", sep=";", encoding="latin1"
)
vagas10 = pd.read_csv(
    RAW_DATA_PATH / "consulta_vagas_2010_BRASIL.csv", sep=";", encoding="latin1"
)
vagas14 = pd.read_csv(
    RAW_DATA_PATH / "consulta_vagas_2014_BRASIL.csv", sep=";", encoding="latin1"
)
vagas18 = pd.read_csv(
    RAW_DATA_PATH / "consulta_vagas_2018_BRASIL.csv", sep=";", encoding="latin1"
)
vagas22 = pd.read_csv(
    RAW_DATA_PATH / "consulta_vagas_2022_BRASIL.csv", sep=";", encoding="latin1"
)

vagas = pd.concat([vagas98, vagas02, vagas06, vagas10, vagas14, vagas18, vagas22], axis=0)

vagas['qt_vaga'] = vagas[['QT_VAGAS', 'QT_VAGA']].bfill(axis=1).iloc[:, 0]
vagas = vagas.drop(columns=['QT_VAGAS', 'QT_VAGA'])
vagas.columns = vagas.columns.str.lower()

vagas = vagas[
    [
        "ano_eleicao",
        "sg_uf",
        "ds_cargo",
        "qt_vaga",
    ]
]

vagas['qt_vaga'] = vagas['qt_vaga'].fillna(0).astype(int)

vagas["ds_cargo"] = vagas["ds_cargo"].str.upper()

vagas = vagas.loc[vagas["ds_cargo"] == "DEPUTADO FEDERAL"].copy()

vagas.to_parquet(
    PROCESSED_DATA_PATH / "vagas_deputado_federal.parquet",
    index=False,
    engine="pyarrow",
)
```
